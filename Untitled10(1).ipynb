```python
for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(column, ":", len(outliers), "outliers")
```



"I identified statistical outliers using the IQR method. Rather than automatically deleting them, I retained them because extreme environmental conditions can be meaningful signals for crop suitability."

```python
X = df.drop("crop_label", axis=1)
```

```python
y = df["crop_label"]
```

```python
numeric_features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

categorical_features = [
    "soil"
]
```

```python
#Encoding Soil Texture
OneHotEncoder(handle_unknown="ignore")
```

```python
StandardScaler()
```

for the numerical variables.

This transforms variables approximately according to:

$$ z = \frac{x-\mu}{\sigma} $$

This is especially useful for algorithms such as:

Logistic Regression
SVM
Neural Networks

Random Forest itself does not require scaling, but using the same preprocessing pipeline makes the project cleaner and easier to deploy.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

preprocessor = ColumnTransformer(
    transformers=[
        
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            numeric_features
        ),

        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_features
        )
    ]
)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Why stratify=y?

Because we have 10 crop categories.

Stratification ensures that the training and testing datasets preserve approximately the same crop proportion

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = Pipeline([
    (
        "preprocessing",
        preprocessor
    ),

    (
        "classifier",
        RandomForestClassifier(
            n_estimators=150,
            random_state=42,
            class_weight="balanced"
        )
    )
])
```

```python
rf_model.fit(X_train, y_train)
```

```python
rf_predictions = rf_model.predict(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = Pipeline([
    (
        "preprocessing",
        preprocessor
    ),

    (
        "classifier",
        LogisticRegression(
            max_iter=1000
        )
    )
])
```

```python
lr_model.fit(X_train, y_train)
```

```python
lr_predictions = lr_model.predict(X_test)
```

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

Why?

Instead of relying on only one train/test split, we divide the training data into five stratified folds.

Each fold maintains representation of the 10 crop classes.

```python
#Benchmark Both Models

from sklearn.model_selection import cross_validate

models = {
    "Random Forest": rf_model,
    "Logistic Regression": lr_model
}

results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=[
            "accuracy",
            "precision_macro",
            "recall_macro",
            "f1_macro",
            "precision_weighted",
            "recall_weighted",
            "f1_weighted"
        ]
    )

    results.append({
        "Model": name,

        "Accuracy":
            scores["test_accuracy"].mean(),

        "Macro Precision":
            scores["test_precision_macro"].mean(),

        "Macro Recall":
            scores["test_recall_macro"].mean(),

        "Macro F1":
            scores["test_f1_macro"].mean(),

        "Weighted Precision":
            scores["test_precision_weighted"].mean(),

        "Weighted Recall":
            scores["test_recall_weighted"].mean(),

        "Weighted F1":
            scores["test_f1_weighted"].mean()
    })

results_df = pd.DataFrame(results)

print(results_df)

```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        rf_predictions
    )
)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    rf_predictions,
    labels=sorted(y.unique())
)
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=sorted(y.unique()),
    yticklabels=sorted(y.unique())
)

plt.xlabel("Predicted Crop")
plt.ylabel("Actual Crop")
plt.title("Crop Recommendation Confusion Matrix")

plt.show()
```

Interpretation

For the supplied dataset, the confusion matrix is effectively diagonal.

That means the model correctly separates the crop categories in the test data.

```python
rf_model.fit(X_train, y_train)

feature_names = (
    rf_model
    .named_steps["preprocessing"]
    .get_feature_names_out()
)

importance = (
    rf_model
    .named_steps["classifier"]
    .feature_importances_
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

print(feature_importance.head(15))
```

```python
plt.figure(figsize=(10,6))

top_features = feature_importance.head(15)

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title("Top Feature Importances")

plt.show()
```

"The Random Forest feature importance analysis helps us understand which soil and environmental measurements contribute most strongly to the crop classification decision. The importance values indicate predictive contribution, not causality."
